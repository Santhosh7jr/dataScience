```python
def design_image():

  path = input("Enter The Image Path: ")

  image = cv2.imread(path)

  if image is None:
    raise Exception("Incorrect Image Path!")

  option = int(input("Enter 1 -> Draw a Line \n 2 -> Draw a Circle \n 3 -> Draw a Rectangle \n Draw a Text: "))

  if option == 1:

    user_input = [int(i.strip()) for i in input("Enter x1, y1, x2, y2, blue_value, green_value, red_value, thickness in comma seperated format: ").split(",")]

    cv2.line(image, (user_input[0], user_input[1]), (user_input[2], user_input[3]), (user_input[4], user_input[5], user_input[6]), user_input[7])

    cv2.namedWindow("Image", cv2.WINDOW_NORMAL)
    cv2.imshow("Image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

  elif option == 2:

    user_input = [int(i.strip()) for i in input("Enter x, y, radius, blue_value, green_value, red_value, thickness in comma seperated format: ").split(",")]
    
    cv2.circle(image, (user_input[0], user_input[1]), user_input[2], (user_input[3], user_input[4], user_input[5]), user_input[6])
    
    cv2.namedWindow("Image", cv2.WINDOW_NORMAL)
    cv2.imshow("Image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

  elif option == 3:
  
      user_input = [int(i.strip()) for i in input("Enter x1, y1, x2, y2, x3, y3, x4, y4, blue_value, green_value, red_value, thickness in comma seperated format: ").split(",")]
      
      cv2.line(image, (user_input[0], user_input[1]), (user_input[2], user_input[3]), user_input[8], user_input[9])
      cv2.line(image, (user_input[2], user_input[3]), (user_input[4], user_input[5]), user_input[8], user_input[9])
      cv2.line(image, (user_input[4], user_input[5]), (user_input[6], user_input[7]), user_input[8], user_input[9])
      cv2.line(image, (user_input[6], user_input[7]), (user_input[0], user_input[1]), user_input[8], user_input[9])
      
      cv2.namedWindow("Image", cv2.WINDOW_NORMAL)
      cv2.imshow("Image", image)
      cv2.waitKey(0)
      cv2.destroyAllWindows()

  elif option == 4:

      user_input = [i.strip() for i in input("Enter text, x, y, font_size, blue_value, green_value, red_value, thickness in comma seperated format: ").split(",")]

      print(user_input)
          
      cv2.putText(image, user_input[0], (int(user_input[1]), int(user_input[2])), cv2.FONT_HERSHEY_PLAIN, int(user_input[3]), (int(user_input[4]), int(user_input[5]), int(user_input[6])), int(user_input[7]))
          
      cv2.namedWindow("Image", cv2.WINDOW_NORMAL)
      cv2.imshow("Image", image)
      cv2.waitKey(0)
      cv2.destroyAllWindows()
  

design_image()
```
