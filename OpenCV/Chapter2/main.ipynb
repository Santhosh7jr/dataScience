```python
import cv2
```

```python
img = cv2.imread("F:/Data_Science_With_Krish_Naik/OpenCV/images/img3.jfif")
```

```python
img.shape
```

```python
grey = cv2.cvtColor(img[:900, :], cv2.COLOR_RGB2GRAY)
```

```python
cv2.imshow("Hairstyle", grey)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imwrite("F:/Data_Science_With_Krish_Naik/OpenCV/saved_images/img3.png", grey)
```

```python
height, width = img.shape[:2]
center = (width // 2, height // 2)

M = cv2.getRotationMatrix2D(center, -90, 1.0)
rotated = cv2.warpAffine(img, M, (width, height))

cv2.namedWindow("Image", cv2.WINDOW_NORMAL)
cv2.imshow("Image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
