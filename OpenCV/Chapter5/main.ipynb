```python
import cv2
import numpy as np
```

```python
image = cv2.imread("F:/Data_Science_With_Krish_Naik/OpenCV/saved_images/img3.png")

blurred = cv2.GaussianBlur(image, (7, 7), 2)

cv2.imshow("original image", image)
cv2.imshow("blurred image", blurred)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blurred = cv2.medianBlur(image, 7)

cv2.imshow("clean image", blurred)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
sharpened = cv2.filter2D(blurred, -1, np.array([
  [0, -1, 0],
  [-1, 5, -1],
  [0, -1, 0]
]))

cv2.imshow("enhanced image", sharpened)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
