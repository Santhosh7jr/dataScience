## Circles

```python
diameter = 10
radius = diameter / 2

print (f"circumference of a circle with radius {radius} is: {2 * np.pi * radius}")
print (f"area of a circle with radius {radius} is: {np.pi * radius * radius}")
print (f"circumference of a circle with radius {radius} is: {2 * np.pi * radius}")
```

## Volumes

```python
length = 5
width = 4
height = 6
radius = 3
```

```python
print (f"Volume of a rectangle with length {length}, width {width} and height {height} is: {length * width * height}")
print (f"Volume of a square with sidelength {length} is: {length * length * length}")
print (f"Volume of a cylinder with length {length} and radius {radius} is: {np.round(np.pi * radius * radius * height, 2)}")
print (f"Volume of a sphere with radius {radius} is: {(4/3) * np.pi * radius * radius * radius}")
print (f"Volume of a cone with height {height} and radius {radius} is: {np.round((1/3) * np.pi * radius * radius * height, 2)}")
```

## Pythagorean Equation

```python
a = 3
b = 4

c = np.round(np.sqrt(a ** 2 + b ** 2), 2)
c
```

## distance formula

```python
point_p1 = [1, 3]
point_p2 = [4, 1]

distance = np.round(np.sqrt(((point_p2[0] - point_p1[0]) ** 2) + ((point_p2[1] - point_p1[1]) ** 2)), 2)
distance
```
