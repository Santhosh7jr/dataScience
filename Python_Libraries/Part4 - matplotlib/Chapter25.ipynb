```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), '--')
plt.grid()
```

```python
plt.figure()

plt.subplot(2, 1, 1)
plt.plot(x, np.cos(x))
```

```python
fig, ax = plt.subplots(2)

ax[0].plot(x, np.sin(x))
ax[1].plot(x, np.cos(x))
```
