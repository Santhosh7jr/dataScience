## Missing Data in Pandas

```python
vals1 = np.array([1, 2, True, 2.5])
vals1
```

```python
vals1 = np.array([1, 2, True, 2.5, None])
vals1
```

```python
vals2.sum()
```

## Operating on Null Values

### isnull
Generates a Boolean mask indicating missing values
### notnull
Opposite of isnull
### dropna
Returns a filtered version of the data
### fillna
Returns a copy of the data with missing values filled or imputed

## Detecting Null Values

```python
data = pd.Series([1, np.nan, "hello", None])
data.isnull()
```

```python
data.notnull()
```

```python
data[data.notnull()]
```

```python
data.dropna()
```

```python
df = pd.DataFrame([
    [1, 2, np.nan],
    [3, 6, 4],
    [np.nan, 5, 6]
])

df
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df2 = pd.DataFrame([
    [1, 2, np.nan],
    [3, np.nan, np.nan],
    [np.nan, 4, np.nan]
])

df2
```

```python
df2.dropna(axis='columns', how='any')
```

```python
df2.dropna(axis='columns', how='all')
```

```python
df2.dropna(axis='rows', thresh=2)
```

## Filling Null Values

```python
data = pd.Series([1, np.nan, 2, None, 3], index=list('abcde'), dtype='Int32')
data
```

```python
data.fillna(0)
```

```python
data.fillna(method="ffill")
```

```python
data.fillna(method='bfill')
```

```python
df = pd.DataFrame([
    [1, 2, np.nan, np.nan],
    [3, np.nan, 4, np.nan],
    [np.nan, 5, 6, np.nan]
])

df
```

```python
df.fillna(0)
```

```python
print(df,"\n")

df.fillna(method='ffill')
```

```python
df.fillna(method='ffill', axis=1)
```
