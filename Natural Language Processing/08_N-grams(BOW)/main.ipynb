```python
import pandas as pd
```

```python
df = pd.read_csv("SMSSpamCollection", sep='\t', names=['label', 'message'])
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
corpus = []

for i in range(len(df)):

  corpus.append(" ".join([lemmatizer.lemmatize(word, pos='v') for word in re.sub('[^a-zA-Z]', ' ', df['message'][i]).lower().split() if word not in stopwords.words('english')]))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=100, binary=True, ngram_range=(1, 2))
```

```python
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
```

```python
cv2 = CountVectorizer(max_features=500, binary=True, ngram_range=(2, 2))
X2 = cv2.fit_transform(corpus).toarray()
cv2.vocabulary_
```

```python
cv2 = CountVectorizer(max_features=500, binary=True, ngram_range=(2, 3))
X2 = cv2.fit_transform(corpus).toarray()
cv2.vocabulary_
```
