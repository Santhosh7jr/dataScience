## Generators

```python
def square(n):
    for i in range(1, n + 1):
        return i ** 2
    
print(square(5))
```

```python
def square(n):
    for i in range(1, n + 1):
        yield i ** 2

print(square(5))
```

```python
generator = square(5)
```

```python
next(generator)
```

```python
def readFile(filename):
    with open(filename, 'r') as file:
        for line in file.readlines():
            yield line

object = readFile("sample.txt")
```

```python
for line in object:
    print(line)
```

# Function Copy

```python
def function():
    return "hello, World!"
```

```python
func = function
print(function())
print(func())

del function

try:
    print(function())
except NameError:
    print("function \"functoin\" is not defined")

print(func())

del func

try:
    print(func())
except NameError:
    print("function \"func\" is not defined")
```

# Closures

```python
def main_method(msg1):

    def sub_method(func, msg2):
        func(msg1)
        return msg2
    
    return sub_method(print, "this is an amazing cource")

print(main_method("welcome to this cource"))
```

# Decorators

```python
def decorator(function):

    def wrapper():

        print("hey guys!")
        function()
        print("I'm a Wrapper")
    
    return wrapper
```

```python
def function():
    print("I am an independent function")
```

```python
function()
```

```python
decorator(function)
```

```python
@decorator
def function():
    print("I am an independent function")
```

```python
function()
```
